## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Looking at the raw data, impression_tier has four levels: low, moderate, good, excellent. Most pages appear to cluster in the lower tiers, this is the crowding problem I want to quantify. I'll use this column and position_tier to see which signal combinations appear in high-visibility pages versus low-visibility ones.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

#====1.How crowded is visibility? 
tier_order = ['low', 'moderate', 'good', 'excellent']
tier_count = df['impression_tier'].value_counts()
tier_pct = (tier_count / len(df) * 100).round(1)
print("Impression tier distbution")
result = pd.DataFrame({'count': tier_count, 'pct': tier_pct})
print(result)
#====2. Median CTR by position tier (filter out avg_position == 0)
has_pos = df[df['avg_position'] > 0]
result = has_pos.groupby('position_tier')['ctr'].median().sort_values(ascending=False)
print("Median CTR by position tier:")
print(result)


#====3. How many pages are 'good' impression tier but still declining?
goodAndDeclining = df[(df['impression_tier'] == 'good') & (df['trend_direction'] == 'down')]

print(f"\nPages with 'good' impressions that are STILL declining: {len(goodAndDeclining):,}")
print(f"That's {len(goodAndDeclining)/len(df[df['impression_tier']=='good']):.1%} of all 'good' impression pages")
```

<h2>Three numbers from the starter dataset support studying ranking signals:</h2>

1. Visibility is highly concentrated :  72.4% of the 30,000 pages in this dataset sit at low or moderate impression tiers, meaning most content gets little to no search exposure. Only 3.6% reach excellent. The crowding problem is real and measurable.
2. Position tier is the strongest observable CTR signal: median CTR at page_1 (0.16%) is roughly 5× higher than at page_3_5 (0.03%), and near zero for deeper positions. This is not surprising, but quantifying it establishes position as the dominant signal to account for in any analysis.
3. High visibility does not guarantee stability: 58.6% of pages with 'good' impressions are still declining in trend. This means reaching visibility is necessary but not sufficient.
